```python

# ============================================================
# PROJECT 06: MARKETING ATTRIBUTION & CAMPAIGN PERFORMANCE
# B.Tech 1st Year - Python Data Analytics Project
# ============================================================

# 1. IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

print("=" * 60)
print("MARKETING ATTRIBUTION & CAMPAIGN PERFORMANCE")
print("=" * 60)

# 2. CREATE SIMULATED MARKETING DATASET

channels = ["Google Ads", "Instagram", "Facebook", "Email", "YouTube"]
audiences = ["Students", "Professionals", "General"]

rows = []
dates = pd.date_range("2026-01-01", periods=90)

for date in dates:
    for channel in channels:

        spend = np.random.randint(500, 3000)
        impressions = np.random.randint(5000, 20000)
        clicks = int(impressions * np.random.uniform(0.02, 0.08))
        sessions = int(clicks * np.random.uniform(0.75, 0.95))
        leads = int(sessions * np.random.uniform(0.10, 0.25))
        conversions = int(leads * np.random.uniform(0.10, 0.35))
        orders = conversions
        revenue = orders * np.random.randint(800, 2000)
        audience = np.random.choice(audiences)

        rows.append([
            date, channel, audience, spend, impressions,
            clicks, sessions, leads, conversions, orders, revenue
        ])

df = pd.DataFrame(rows, columns=[
    "Date", "Channel", "Audience", "Spend", "Impressions",
    "Clicks", "Sessions", "Leads", "Conversions", "Orders", "Revenue"
])

print("\n1. DATASET CREATED")
print(df.head())
print("Dataset Shape:", df.shape)

# 3. DATA CLEANING AND VALIDATION

df.drop_duplicates(inplace=True)
df["Date"] = pd.to_datetime(df["Date"])
df["Channel"] = df["Channel"].str.strip().str.title()

print("\n2. DATA CLEANING")
print("Missing Values:\n", df.isnull().sum())

assert (df["Clicks"] <= df["Impressions"]).all()
assert (df["Sessions"] <= df["Clicks"]).all()
assert (df["Leads"] <= df["Sessions"]).all()
assert (df["Conversions"] <= df["Leads"]).all()

print("Data validation successful!")

# 4. CAMPAIGN DATA RECONCILIATION

print("\n3. CAMPAIGN DATA RECONCILIATION")

for col in ["Spend", "Impressions", "Clicks", "Sessions",
            "Leads", "Conversions", "Orders", "Revenue"]:
    print("Total", col + ":", df[col].sum())

# 5. CALCULATE MARKETING KPIs

summary = df.groupby("Channel").agg({
    "Spend": "sum",
    "Impressions": "sum",
    "Clicks": "sum",
    "Sessions": "sum",
    "Leads": "sum",
    "Conversions": "sum",
    "Orders": "sum",
    "Revenue": "sum"
}).reset_index()

summary["CTR (%)"] = (
    summary["Clicks"] / summary["Impressions"] * 100
)

summary["CPC"] = (
    summary["Spend"] / summary["Clicks"]
)

summary["CVR (%)"] = (
    summary["Conversions"] / summary["Sessions"] * 100
)

summary["CPA"] = (
    summary["Spend"] / summary["Conversions"].replace(0, np.nan)
)

summary["ROAS"] = (
    summary["Revenue"] / summary["Spend"]
)

summary["Spend Contribution (%)"] = (
    summary["Spend"] / summary["Spend"].sum() * 100
)

summary["Revenue Contribution (%)"] = (
    summary["Revenue"] / summary["Revenue"].sum() * 100
)

print("\n4. MARKETING KPI REPORT")
print(summary.round(2).to_string(index=False))

# 6. NUMPY UNCERTAINTY ANALYSIS

sessions = df["Sessions"].to_numpy()
conversions = df["Conversions"].to_numpy()

observed_cvr = conversions.sum() / sessions.sum()

bootstrap_results = []

for i in range(1000):

    sample = np.random.randint(0, len(df), len(df))

    sampled_sessions = sessions[sample]
    sampled_conversions = conversions[sample]

    cvr = sampled_conversions.sum() / sampled_sessions.sum()

    bootstrap_results.append(cvr * 100)

lower = np.percentile(bootstrap_results, 2.5)
upper = np.percentile(bootstrap_results, 97.5)

print("\n5. NUMPY UNCERTAINTY ANALYSIS")

print("Average Conversion Rate:",
      round(observed_cvr * 100, 2), "%")

print("95% Confidence Interval:",
      round(lower, 2), "% to", round(upper, 2), "%")

roas_values = summary["ROAS"].to_numpy()

print("Average ROAS:", round(np.mean(roas_values), 2))
print("Maximum ROAS:", round(np.max(roas_values), 2))
print("Minimum ROAS:", round(np.min(roas_values), 2))
print("ROAS Standard Deviation:", round(np.std(roas_values), 2))

best_channel = summary.loc[
    summary["ROAS"].idxmax(), "Channel"
]

print("Best Channel by ROAS:", best_channel)

# 7. MONTHLY AND AUDIENCE PERFORMANCE

df["Month"] = df["Date"].dt.month_name()

monthly_summary = df.groupby("Month").agg({
    "Spend": "sum",
    "Clicks": "sum",
    "Conversions": "sum",
    "Revenue": "sum"
})

monthly_summary["ROAS"] = (
    monthly_summary["Revenue"] / monthly_summary["Spend"]
)

audience_summary = df.groupby("Audience").agg({
    "Spend": "sum",
    "Clicks": "sum",
    "Conversions": "sum",
    "Revenue": "sum"
})

audience_summary["ROAS"] = (
    audience_summary["Revenue"] / audience_summary["Spend"]
)

print("\n6. MONTHLY PERFORMANCE")
print(monthly_summary.round(2))

print("\nAUDIENCE PERFORMANCE")
print(audience_summary.round(2))

# 8. SIMULATE CUSTOMER JOURNEYS

journeys = []

for i in range(1, 501):

    number_of_touchpoints = np.random.randint(1, 4)

    touchpoints = list(
        np.random.choice(
            channels,
            size=number_of_touchpoints,
            replace=True
        )
    )

    journeys.append({
        "Order_ID": i,
        "Touchpoints": touchpoints
    })

journey_df = pd.DataFrame(journeys)

# 9. ATTRIBUTION MODEL COMPARISON

first_touch = {c: 0 for c in channels}
last_touch = {c: 0 for c in channels}
linear_touch = {c: 0 for c in channels}

for journey in journeys:

    touchpoints = journey["Touchpoints"]

    first_touch[touchpoints[0]] += 1
    last_touch[touchpoints[-1]] += 1

    for channel in touchpoints:
        linear_touch[channel] += 1 / len(touchpoints)

attribution = pd.DataFrame({
    "Channel": channels,
    "First Touch": [first_touch[c] for c in channels],
    "Last Touch": [last_touch[c] for c in channels],
    "Linear Touch": [linear_touch[c] for c in channels]
})

print("\n7. ATTRIBUTION MODEL COMPARISON")
print(attribution.round(2).to_string(index=False))

# 10. BUDGET REALLOCATION SIMULATION

total_budget = summary["Spend"].sum()

allocation = summary.set_index("Channel")["Spend"].astype(float).copy()

best = summary.loc[
    summary["ROAS"].idxmax(), "Channel"
]

worst = summary.loc[
    summary["ROAS"].idxmin(), "Channel"
]

minimum_budget = total_budget * 0.10
maximum_budget = total_budget * 0.35

transfer = max(0, min(
    total_budget * 0.10,
    allocation[worst] - minimum_budget,
    maximum_budget - allocation[best]
))

allocation[worst] -= transfer
allocation[best] += transfer

roas_dict = summary.set_index("Channel")["ROAS"]

estimated_revenue = allocation * roas_dict

print("\n8. BUDGET REALLOCATION")

print("Original Budget:", round(total_budget, 2))
print("New Budget:", round(allocation.sum(), 2))

print("\nReallocated Budget:")
print(allocation.round(2))

print("\nEstimated Revenue:")
print(estimated_revenue.round(2))

print("Budget shifted from:", worst)
print("Budget shifted to:", best)

print("Original Revenue:",
      round(summary["Revenue"].sum(), 2))

print("Scenario Revenue:",
      round(estimated_revenue.sum(), 2))

print("Note: Scenario assumes historical ROAS remains constant.")

# 11. DATA VISUALISATION

# Graph 1: Marketing Spend

plt.figure(figsize=(9, 5))
plt.bar(summary["Channel"], summary["Spend"])

plt.title("Marketing Spend by Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("Total Spend")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# Graph 2: ROAS Comparison

plt.figure(figsize=(9, 5))
plt.bar(summary["Channel"], summary["ROAS"])

plt.title("ROAS Comparison by Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("ROAS")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# Graph 3: Conversions

plt.figure(figsize=(9, 5))
plt.bar(summary["Channel"], summary["Conversions"])

plt.title("Conversions by Marketing Channel")
plt.xlabel("Channel")
plt.ylabel("Conversions")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# Graph 4: Attribution Comparison

attribution.plot(
    x="Channel",
    y=["First Touch", "Last Touch", "Linear Touch"],
    kind="bar",
    figsize=(10, 5)
)

plt.title("Attribution Model Comparison")
plt.xlabel("Marketing Channel")
plt.ylabel("Attributed Conversion Credit")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# Graph 5: Monthly Revenue

monthly_summary["Revenue"].plot(
    kind="line",
    marker="o",
    figsize=(9, 5)
)

plt.title("Monthly Revenue Performance")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)
plt.tight_layout()
plt.show()

# 12. SAVE ALL PROJECT OUTPUTS

df.to_csv("marketing_campaign_data.csv", index=False)

summary.to_csv("campaign_kpi_report.csv", index=False)

attribution.to_csv("attribution_comparison.csv", index=False)

monthly_summary.to_csv("monthly_performance.csv")

audience_summary.to_csv("audience_performance.csv")

allocation.rename("Allocated Budget").to_csv(
    "budget_allocation.csv"
)

journey_df.to_csv("customer_journeys.csv", index=False)

# 13. FINAL PROJECT SUMMARY

print("\n" + "=" * 60)
print("PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 60)

print("\nProject Summary:")
print("1. Marketing dataset created.")
print("2. Data cleaned and validated.")
print("3. Marketing KPIs calculated.")
print("4. NumPy uncertainty analysis completed.")
print("5. Monthly and audience summaries created.")
print("6. Attribution models compared.")
print("7. Budget reallocation simulated.")
print("8. Data visualisations generated.")
print("9. CSV reports exported.")

print("\nBest Channel by Observed ROAS:", best_channel)

print("\nAll output CSV files are saved successfully!")
print("\nTHANK YOU!")
```
